## CODECHECKER notes

A few dependencies had to be manually installed on Ubuntu Linux so that renv succeeded in
installing the R environment. Authors helped to identify the issues, and commented they
would address the installation instructions. The processing time was also not documented,
but it matched the time informed by authors afterwards, of approximately eleven hours. The
output of the R Quarto document produce violin plots that seemed a bit different from those
from the paper. However, executing the rest of the Jupyter notebook, the same plots are
produced in Python and these look identical to the ones from the paper. No other issues
found.

## Recommendations to the authors

- Update the installation instructions;
- Mention that some files are pre-generated with a different version from the final used in the paper;
- Include a continuous integration test pipeline, so readers/users can use that as reference.
